```python
import csv
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Step 1: Load data from CSV file and preprocess
x = []
y = []

with open("data_new.csv", "r") as csv_file:
    csv_reader = csv.reader(csv_file, delimiter=",")
    convert_rgb = lambda color: tuple(int(color[i:i+2], 16) for i in (0,2,4))
    for row in csv_reader:
        x.append(convert_rgb(row[0]))
        y.append([convert_rgb(x) for x in row[1].split(" ")])
        
train_x = np.array(x)
train_y = np.array(y)

# Step 2: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(train_x, train_y, train_size=0.95)

# Reshape y_train to have two dimensions
n_samples, n_colors, n_channels = y_train.shape
y_train_reshaped = y_train.reshape(n_samples, n_colors * n_channels)
```

```python
# Train the linear regression model
model = LinearRegression()
model.fit(X_train, y_train_reshaped)
```

```python
# Step 4: Evaluate the model
y_pred = model.predict(X_test)
# Reshape y_test to match the shape of y_pred
print(y_test.shape)
n_samples_test, n_colors_test, n_channels_test = y_test.shape
y_test_reshaped = y_test.reshape(n_samples_test, n_colors_test * n_channels_test)
print(y_test_reshaped.shape)
mse = mean_squared_error(y_test_reshaped, y_pred)
print("Mean Squared Error:", mse)
```
